```python
sns.scatterplot(data=df, x='IMDbRating', y='RottenTomatoesScore', alpha = 0.5)
plt.title("IMDb Rating vs Rotten Tomatoes Score")
plt.xscale('log')
plt.yscale('log')
plt.show()
```

```python
sns.scatterplot(data=df, x='BudgetUSD', y='Global_BoxOfficeUSD', alpha = 0.5)
plt.title("Budget vs Global Box Office")
plt.xscale('log')
plt.yscale('log')
plt.show()
```

```python
sns.scatterplot(data=df, x='BudgetUSD', y='Global_BoxOfficeUSD', alpha = 0.5)
plt.title("Budget vs Global Box Office")
plt.xscale('linear')
plt.yscale('linear')
plt.show()
```

```python
genre_year_stats = df.groupby(["Genre", "ReleaseYear"]).agg({"Global_BoxOfficeUSD": "mean", "BudgetUSD": "mean", "IMDbRating": "mean", "RottenTomatoesScore": "mean"}).reset_index()
```

```python
genre_year_stats
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=genre_year_stats, x='ReleaseYear', y='Global_BoxOfficeUSD', hue='Genre')
plt.title("Average Global Box Office by Genre Over Years")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=genre_year_stats.query("Genre == 'Action'"), x='ReleaseYear', y='Global_BoxOfficeUSD', hue='Genre')
plt.title("Average Global Box Office by Genre Over Years")
plt.show()
```

```python
top_actors = df.groupby("LeadActor")["IMDbRating"].mean().sort_values(ascending=False).head(10)
```

```python
top_actors
```

```python
top_actors.plot(kind='bar', figsize=(12, 6))
plt.title("Top 10 Lead Actors by Average IMDb Rating")
plt.show()
```
